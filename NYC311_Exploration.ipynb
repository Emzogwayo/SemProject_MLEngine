# Dataset Candidates and Selection

## Dataset Candidates Explored

The following open-source datasets were considered as possible datasets for the Madaraka Estate semester project:

1. **NYC 311 Service Requests**
   Contains service requests made by residents about different municipal issues.
   **Expected Output (Target Variable):** `Complaint Type`

2. **Chicago 311 Service Requests**
   Contains service requests relating to different issues reported across the city.
   **Expected Output (Target Variable):** `SR_TYPE`

3. **Kaggle Civic Issue Reports**
   Contains citizen-reported infrastructure issues and their descriptions.
   **Expected Output (Target Variable):** `Priority_Level`

## Selected Dataset

We selected the **NYC 311 Service Requests dataset** because it contains different types of community service requests that are similar to the types of issues that may be reported in Madaraka Estate, such as water, electricity and waste-related problems.

The `Complaint Type` variable can be used as the target variable for a classification model. The aim would be to use information from a resident's report to predict the type of issue being reported. This could later help the Madaraka Estate system classify and direct reports more efficiently.
